```python
from datasets import load_dataset

dataset = load_dataset("stanfordnlp/imdb")
```

```python
train_df = pd.DataFrame(dataset["train"])
test_df = pd.DataFrame(dataset["test"])

print("Training data:")
print(train_df.head())
```

```python
print("Rows and columns:", train_df.shape)
print("\nColumn names:")
print(train_df.columns)

print("\nData types:")
print(train_df.dtypes)
```

```python
print("Missing values:")
print(train_df.isnull().sum())
```

```python
print("Duplicate rows:", train_df.duplicated().sum())
```

```python
before = len(train_df)

train_df = train_df.drop_duplicates()

after = len(train_df)

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Duplicates removed:", before - after)
```

```python
print("Review sentiment count:")
print(train_df["label"].value_counts())
```

```python
positive = (train_df["label"] == 1).sum()
negative = (train_df["label"] == 0).sum()
total = len(train_df)

print("Positive reviews:", positive)
print("Negative reviews:", negative)
print("Total reviews:", total)

print("Positive percentage:", (positive / total) * 100)
print("Negative percentage:", (negative / total) * 100)
```

```python
labels = np.array(train_df["label"])

print("NumPy array:")
print(labels[:20])
```

```python
print("Mean:", np.mean(labels))
print("Median:", np.median(labels))
print("Minimum:", np.min(labels))
print("Maximum:", np.max(labels))
print("Standard deviation:", np.std(labels))
```

```python
train_df["review_length"] = train_df["text"].str.len()

print(train_df[["text", "review_length"]].head())
```

```python
lengths = np.array(train_df["review_length"])

print("Average review length:", np.mean(lengths))
print("Median review length:", np.median(lengths))
print("Shortest review:", np.min(lengths))
print("Longest review:", np.max(lengths))
print("Standard deviation:", np.std(lengths))
```

```python
positive_lengths = train_df[train_df["label"] == 1]["review_length"]
negative_lengths = train_df[train_df["label"] == 0]["review_length"]

print("Average positive review length:", np.mean(positive_lengths))
print("Average negative review length:", np.mean(negative_lengths))
```

```python
difference = np.mean(positive_lengths) - np.mean(negative_lengths)

print("Difference in average length:", difference)
```

```python
import matplotlib.pyplot as plt

sentiments = ["Positive", "Negative"]
counts = [positive, negative]

plt.bar(sentiments, counts)

plt.title("Positive vs Negative Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
import matplotlib.pyplot as plt

sentiments = ["Positive", "Negative"]
counts = [positive, negative]

plt.bar(sentiments, counts)

plt.title("Positive vs Negative Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

# Add numbers on top of bars
for i, count in enumerate(counts):
    plt.text(i, count + 100, str(count), ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt

sentiments = ["Positive", "Negative"]
average_lengths = [
    np.mean(positive_lengths),
    np.mean(negative_lengths)
]

plt.bar(sentiments, average_lengths)

plt.title("Average Review Length by Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Average Review Length (Characters)")

# Add values on top of bars
for i, value in enumerate(average_lengths):
    plt.text(i, value + 20, f"{value:.2f}", ha="center")

plt.show()
```
